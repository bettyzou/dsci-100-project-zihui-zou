#### Task 5.2 Perform the Analysis (20 points)

```r
### EDIT THIS CODE CELL ###
library(tidymodels)
library(tidyverse)

# Make the random split reproducible.
set.seed(123)

# Split the data before imputaion and standardization.
basketball_split <- initial_split(basketball_tidy, prop = 0.75)
basketball_train <- training(basketball_split)
basketball_test <- testing(basketball_split)

# Create five folds using only the training data.
basketball_folds <- vfold_cv(basketball_train, v = 5)
```

```r
# Replace missing predictor values with training means,
# then center and scale both numeric predictors.
basketball_recipe <- recipe(
    tourney_finish ~ seed + conf_wins_pct,
    data = basketball_train
) |>
    step_impute_mean(all_numeric_predictors()) |>
    step_center(all_numeric_predictors()) |>
    step_scale(all_numeric_predictors())

# Tune the number of neighbor using cross-validation.
knn_spec <- nearest_neighbor(
    weight_func = "rectangular",
    neighbors = tune()
) |>
    set_engine("kknn") |>
    set_mode("classification")

# Combine preprocessing and the model.
basketball_workflow <- workflow() |>
    add_recipe(basketball_recipe) |>
    add_model(knn_spec)
```

```r
# Compare odd K values from 1 to 31.
k_grid <- tibble(neighbors = seq(1, 31, by = 2))
set.seed(123)

# Evaluate each K using mean validation accuracy.
basketball_tuning <- basketball_workflow |>
    tune_grid(
        resamples = basketball_folds,
        grid = k_grid,
        metrics = metric_set(accuracy)
    )

# Display cross-validation results.
basketball_cv_results <- collect_metrics(basketball_tuning)
basketball_cv_results 

# Select K with the highest mean validation accuracy.
best_k <- select_best(basketball_tuning, metric = "accuracy")
best_k
```

```r
# Fit the selected workflow using the full training set.
basketball_final_fit <- basketball_workflow |>
    finalize_workflow(best_k) |>
    fit(data = basketball_train)

# Generate predictions for the test set.
basketball_predictions <- predict(
    basketball_final_fit,
    new_data = basketball_test
) |>
    bind_cols(basketball_test)

# Calculate test accuracy.
basketball_accuracy <- basketball_predictions |>
    accuracy(truth = tourney_finish, estimate = .pred_class)

basketball_accuracy

# Examine correct predictions and errors for each category.
basketball_confusion <- basketball_predictions |>
    conf_mat(truth = tourney_finish, estimate = .pred_class)

basketball_confusion
```

```r
# Identify the most common tournament finish in the training set.
majority_finish <- basketball_train |>
    count(tourney_finish, sort = TRUE) |>
    slice(1) |>
    pull(tourney_finish)

# Calculate baseline accuracy on the same test set.
baseline_accuracy <- mean(
    basketball_test$tourney_finish == majority_finish
)

# Compare KNN accuracy with baseline accuracy.
basketball_comparison <- tibble(
    model = c("KNN", "Majority-class baseline"),
    test_accuracy = c(
        basketball_accuracy$.estimate,
        baseline_accuracy
    )
)
basketball_comparison
```
